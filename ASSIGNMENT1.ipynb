```python
#ASSIGNMENT 6.1 WAP USING FUNCTION THAT ADSS ALL ODD NUMBERS FROM 1 TO N, N IS GIVEN BY THE USER
def add_odd(n):
    tot1=0
    for i in range(1,n+1):
        if(i%2!=0):
             tot1+=i
    return tot1
N=int(input("Enter the number: "))
print("The sum of the odd numbers from 1 to",N,"is:",add_odd(N))
```

```python
#ASSIGNMENT 6.2 WAP TO ADD THE PRIME NUMBERS FROM 1 TO N USING FUNCTION
def add_prime(N):
    tot=0
    if(N<=1):
        return tot
    else:
        for j in range(2,N+1):
            prime=True
            for i in range(2,j):
                  if(j%i ==0):
                     prime=False
                     break
            if(prime==True):
                 tot+=j
    return tot
x=int(input("Enter the number: "))
print("The sum of the prime numbers from 1 to",x,"is:",add_prime(x))
```
