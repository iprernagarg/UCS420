```python
steps=np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
#a
print(np.sum(steps))
#b
print(np.mean(steps))
#c
print(np.max(steps))
print(np.min(steps))
#d
print(np.sum(steps,axis=0))
#e
print(np.sum(steps,axis=1))
#f
maxp=np.unravel_index(np.argmax(steps),steps.shape)
print(maxp)
```

```python
X=np.array([[6,70,3],[5,50,6],[8,80,2],[4,30,8]])
y=np.array([40,65,30,85])
#a
print(X.shape)
print(X.ndim)
#b
trans=X.T
print(trans)
#c
prod=X.T @ X
print(prod)
#d
inv=np.linalg.inv(prod)
print(inv)
#e
OLS=inv @ X.T @ y
print(OLS)
#f
print(OLS[0]," Sleep Coeff")
print(OLS[1]," Activity Coeff")
print(OLS[2]," Stress Coeff")
#g
new_user=np.array([5,40,7])
pred=new_user @ OLS
print(pred)
```
