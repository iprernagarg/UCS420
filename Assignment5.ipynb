```python
#b
x=np.array([1,2,3,4,5,1,2,1,1,1])
val,cnts=np.unique(x,return_counts=True)
mostf=val[np.argmax(cnts)]
idx=np.where(x==mostf)
print(mostf)
print(idx)

y=np.array([1,1,1,2,3,4,2,4,3,3])
val1,cnts1=np.unique(y,return_counts=True)
mostf1=val1[np.argmax(cnts1)]
idx1=np.where(y==mostf1)
print(mostf1)
print(idx1)

```
