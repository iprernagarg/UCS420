```python
import random
roll=1024160066
random.seed(roll)
random_no=[random.randint(100,900) for _ in range(100)]
print("First 10 random numbers:",random_no[:10])
```

```python
odd=sum(1 for n in random_no if n%2 !=0)
even=sum(1 for n in random_no if n%2 ==0)
print("Odd numbers count:",odd)
print("Even numbers count:",even)

```

```python
def is_prime(num):
    if num<2:return False
    for i in range(2,int(num**0.5)+1):
        if num%i==0: return False
    return True
prime=[n for n in random_no if is_prime(n)]
print(len(prime))
print(prime)
    
```

```python
most_freq=max(set(random_no),key=random_no.count)
freq=random_no.count(most_freq)
print(most_freq)
print(freq,"times")
```

```python
roll=[int(d) for d in "86664211"]
A={digit*7 for digit in roll}
B={digit*9 for digit in roll}
print(A)
print(B)
```

```python
print("Union:",A.union(B))
print("Intersection:",A.intersection(B))
```

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
print("Symmetric Difference:", A.symmetric_difference(B))
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value to discard from A: "))
A.discard(X)
# discard() silently does nothing if the element is not found, whereas remove() throws a KeyError.
```

```python
my_dict = {
 "name": "John Doe",
 "roll_no": "1234567890",
 "branch": "Computer Science",
 "age": 20,
 "city": "Patiala"
}

my_dict["location"] = my_dict.pop("city")
print(" Dictionary after renaming city:", my_dict)
```

```python
my_dict["cgpa"] = 9.5
my_dict["age"] += 1
print("After adding cgpa and updating age:", my_dict)
```

```python
copy1 = my_dict.copy()
copy2 = my_dict.copy()
popped_val = copy1.pop("branch")
del copy2["branch"]
# pop() removes the key and returns its value, whereas the del keyword removes the key without returning anything.
```

```python
print("Dictionary Items:")
for key, value in my_dict.items():
    print(f"{key} -> {value}")
```

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Safe fallback: 'email' key is not present in the dictionary.")
```

```python
friend_dict = {
 "name": "Jane Smith",
 "roll_no": "0987654321",
 "branch": "Electrical",
 "age": 21,
 "city": "Chandigarh"
}
merged_dict = {**my_dict, **friend_dict}
# When sharing a key, the values from the dictionary unpacked last (**friend_dict) "win" and overwrite previous values.
```
